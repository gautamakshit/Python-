```python
# Problem 1: Fixing the Buggy Code Super()

class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def display(self):
        print("Name:", self.name)
        print("ID:", self.id)

class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def print_data(self):
        print("Emp class called")

emp_details = Emp("Mayank", 103)

emp_details.display()
emp_details.print_data()
```

```python
# Problem 2: Anagram Chcek

def check_anagram(str1, str2):
    if sorted(str1.lower()) == sorted(str2.lower()):
        print("Strings are Anagrams")
    else:
        print("Strings are Not Anagrams")

s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

check_anagram(s1, s2)
```

```python
# Problem 3: Armstrong Number Check

num = int(input("Enter a number: "))

order = len(str(num))
temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10

if sum == num:
    print(num, "is an Armstrong Number")
else:
    print(num, "is not an Armstrong Number")
```

```python
# problem 4: Classes and objects(vehicle class)
class Vehicle:
    color = "White"

    def __init__(self, max_speed, mileage, seating_capacity):
        self.max_speed = max_speed
        self.mileage = mileage
        self.seating_capacity = seating_capacity

    def display(self):
        print("Color:", Vehicle.color)
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Seating Capacity:", self.seating_capacity)

# Object 1
car1 = Vehicle(200, 20, 5)

# Object 2
car2 = Vehicle(180, 25, 4)

print("Car 1 Details:")
car1.display()

print("\nCar 2 Details:")
car2.display()
```
