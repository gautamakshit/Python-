```python
# Q 4. Check Even/Odd, Palindrome and Divisibility by sum of digits

num = int(input("Enter a number (1-9999): "))

# Even or Odd
if num % 2 == 0:
    print("Even")
else:
    print("Odd")

# Palindrome
if str(num) == str(num)[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

# Divisible by sum of digits
sum_digits = sum(int(d) for d in str(num))

if num % sum_digits == 0:
    print("Divisible by sum of digits")
else:
    print("Not divisible by sum of digits")
```

```python
# Q 7. Copy strings to new list until orange is found

squares = ['red', 'yellow', 'green', 'orange', 'purple', 'blue']

new_squares = []
i = 0

while i < len(squares):
    if squares[i] == 'orange':
        new_squares.append(squares[i])
        break
    new_squares.append(squares[i])
    i += 1

print(new_squares)
```

```python
# Q 8. Animals with names of 7 letters

animals = ["lion", "giraffe", "gorilla", "parrots",
           "crocodile", "deer", "swan"]

seven_letter_animals = []
i = 0

while i < len(animals):
    if len(animals[i]) == 7:
        seven_letter_animals.append(animals[i])
    i += 1

print("Animals with 7 letters:")
print(seven_letter_animals)
```
