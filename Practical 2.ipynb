```python
# Practical 2 - Strings & List

# 1. Create an empty list

shopping_list = []

```

```python
# 2. Store items in shopping list 

shopping_list = ["Watch", "Laptop" , "Shoes" , "Pen" , "Clothes"] 
```

```python
# 3. Add new item Football

shopping_list.append("Football") 
```

```python
# 4. Print Last item

print("shopping_list : ", shopping_list[-1]) 
```

```python
# 5. Print entire shopping list 

print("shopping_list : ",shopping_list) 
```

```python
# 6. Print important items (Laptop and shoes)

print("Important Items:", shopping_list[1],",",
shopping_list[2])
```

```python
# 7. Replace Pen with Notebook 

shopping_list[3] = "Notebook"
print("After Replacing Pen:",shopping_list)
```

```python
# Delete least important item(Watch)

shopping_list.remove("Watch")
print("Final Shopping List:",
      shopping_list)
```
