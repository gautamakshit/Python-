```python
# Problem 4: Data Loading and Exploration

import pandas as pd
import numpy as np

data = {
    'Customer_ID': [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],

    'Name': ['Aman', 'Rahul', 'Priya', 'Neha', 'Rohit',
             'Anjali', 'Karan', 'Simran', 'Vikas', 'Pooja'],

    'Age': [25, 32, 28, 35, 40, 22, 30, 27, 45, 33],

    'Gender': ['Male', 'Male', 'Female', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],

    'Purchase_Amount': [5000, 7000, 4500, 8000, 9000,
                        3000, 6500, np.nan, 7500, 5500]
}

df = pd.DataFrame(data)

# Display first 5 rows
print(df.head())

# Summary statistics
print(df.describe())

# Check missing values
print(df.isnull().sum())

# Handle missing values
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].mean()
)

print(df)
```

```python
# Problem 5 Data Filtering and Selection

# Customers whose age is greater than 30
print(df[df['Age'] > 30])

# Total purchase amount for male customers
male_total = df[df['Gender'] == 'Male']['Purchase_Amount'].sum()

print("Male Purchase Total:", male_total)

# Total purchase amount for female customers
female_total = df[df['Gender'] == 'Female']['Purchase_Amount'].sum()

print("Female Purchase Total:", female_total)

# Select specific columns
selected = df[['Customer_ID', 'Name', 'Purchase_Amount']]

print(selected)
```
